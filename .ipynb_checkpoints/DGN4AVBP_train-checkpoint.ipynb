```python
import os
from Dataset import create_cfd_datamodule,estimate_mesh_scales,_sample_graphs_for_stats
from utils import read_metadata,load_coo_data,create_data_list,create_graph_data
import lightning as L
from dgn4avbp.diffusion_process import DiffusionProcess
from dgn4avbp.dgn_model import DiffusionGraphNet
from dgn4avbp.step_sampler import ImportanceStepSampler
from dgn4avbp.lit_dgn import LitDiffusionCFD
from dgn4avbp.losses import HybridLoss
from lightning.pytorch.callbacks import ModelCheckpoint, RichProgressBar
from lightning.pytorch.callbacks.progress.rich_progress import RichProgressBarTheme
from torchvision import transforms as T
from dgn4avbp.transform_locals import EnsureEdgeAttrFromPos, ScaleEdgeAttr,ScaleAttr,ZScoreTarget
from dgn4avbp.transform_locals import MeshCoarsening  # put your pasted class here
from torch_geometric.data import Data, Batch
import torch

from dgn4avbp.transform_locals import EnsureEdgeAttrFromPos, ScaleEdgeAttr, MeshCoarsening, ZScoreTarget
from torchvision import transforms as T
```

```python
def _sample_graphs_for_stats(meta, K=4):
    pos0, edge_index0, cells0 = load_coo_data(
        meta['coo_file'], meta['mesh_file'], meta['coordinate_paths']
    )
    file_list, it_list_total = create_data_list(
        [meta['solution_directory']],
        meta['seq_len'],
        meta['solution_prefix']
    )
    K = min(K, len(it_list_total))
    samples = []
    for k in range(K):
        case, sim, t = it_list_total[k][0]        # take the first step of each sequence
        fpath = file_list[str(case)][str(t)]
        g = create_graph_data(pos0, edge_index0, fpath, meta, cells0)
        samples.append(g)
    return samples, pos0, edge_index0

def _estimate_mesh_scales(pos, edge_index, quantile=0.5):
    row, col = edge_index
    edge_len = (pos[col] - pos[row]).norm(dim=1)
    h = edge_len.quantile(quantile).item()
    rel_pos_scaling = [h, 2*h, 4*h, 8*h]
    return h, rel_pos_scaling

def get_sequence_from_combined(item):
    # item may be: (batches, batch_idx, dataloader_idx) OR just `batches`
    if isinstance(item, tuple) and len(item) == 3:
        batches, _batch_idx, _dl_idx = item
    else:
        batches = item

    # If CombinedLoader was built as {"main": loader}
    if isinstance(batches, dict):
        sequence = next(iter(batches.values()))
    else:
        sequence = batches

    # Some wrappers yield a tuple rather than list — treat them the same.
    if isinstance(sequence, tuple):
        sequence = list(sequence)
    return sequence  # list[Batch]

def print_batch_info(sequence):
    assert isinstance(sequence, (list, tuple)), f"Expected list/tuple, got {type(sequence)}"
    print(f"    Number of time steps in batch: {len(sequence)}")
    for t, g in enumerate(sequence):
        # If someone tucked the Batch inside {"main": Batch} per time-step:
        if isinstance(g, dict) and "main" in g:
            g = g["main"]
        assert isinstance(g, (Batch, Data)), f"Time step {t} is {type(g)}"
        print(f"    Time step {t+1}:")
        print(f"        Number of graphs in batch: {g.num_graphs}")
        print(f"        Total number of nodes: {g.num_nodes}")
        print(f"        Total number of edges: {g.num_edges}")
        if hasattr(g, 'target'):
            print(f"        target shape: {tuple(g.target.shape)}")
        if hasattr(g, 'edge_attr'):
            print(f"        edge_attr shape: {tuple(g.edge_attr.shape)}")
        if hasattr(g, 'cells') and isinstance(g.cells, list) and g.cells and g.cells[0] is not None:
            print(f"        Total number of cells: {len(g.cells[0])}")
        if hasattr(g, 'time'):
            print(f"        time mean per-graph: {g.time.view(g.num_graphs, -1).mean(dim=-1)}")

class cfd_datamodule(L.LightningDataModule):
    def __init__(self, metadata_files, train_val_split=0.8):
        super().__init__()
        self.metadata_files = metadata_files
        metadata = [read_metadata(mf) for mf in metadata_files]
        self.batch_sizes  = [info['batch_size'] for info in metadata]
        self.loader_types = ['default'] * len(metadata_files)
        self.start_idx    = [0] * len(metadata_files)
        self.split        = train_val_split

        # will be set in setup()
        self.graph_transform = None
        self._zscore_mean = None
        self._zscore_std  = None
        self._rel_pos_scales = None
        self._h = None

    def setup(self, stage: str):
        # --- compute stats ONCE from the first dataset (or loop over all and concat) ---
        meta0 = read_metadata(self.metadata_files[0])
        samples, pos0, edge_index0 = _sample_graphs_for_stats(meta0, K=4)

        # target mean/std (your target has 3 velocity components)
        tgt = torch.cat([g.target for g in samples], dim=0)  # shape (N_total, 3)
        self._zscore_mean = tgt.mean(0)                      # (3,)
        self._zscore_std  = tgt.std(0)                       # (3,)
        print(self._zscore_mean,self._zscore_std)
        # mesh scales for edge normalization & LR scaling
        self._h, self._rel_pos_scales = _estimate_mesh_scales(pos0, edge_index0)

        # --- build the graph transform pipe (reused for train/val/test) ---
        self.graph_transform = T.Compose([
            EnsureEdgeAttrFromPos(),         # edge_attr = pos[j]-pos[i]
            ScaleEdgeAttr(self._h),          # normalize HR edge vectors by median spacing
            ZScoreTarget(self._zscore_mean, self._zscore_std),  # normalize targets
            MeshCoarsening(
                num_scales=4,
                max_indegree=None,           # set later for unstructured meshes
                rel_pos_scaling=None,
                scalar_rel_pos=False,         # keep 3D vectors for LR edges
            ),
        ])

        # --- create loaders with this transform applied in the Collater ---
        if stage == "fit":
            self.train_cfd_datamodule = create_cfd_datamodule(
                self.metadata_files, self.batch_sizes, self.loader_types, self.start_idx,
                shuffle=True, split=self.split, flag='train',
                collater_transform=self.graph_transform
            )
            self.val_cfd_datamodule = create_cfd_datamodule(
                self.metadata_files, self.batch_sizes, self.loader_types, self.start_idx,
                shuffle=False, split=self.split, flag='val',
                collater_transform=self.graph_transform
            )

        if stage == "test":
            self.val_cfd_datamodule = create_cfd_datamodule(
                self.metadata_files, self.batch_sizes, self.loader_types, self.start_idx,
                shuffle=False, split=self.split, flag='val',
                collater_transform=self.graph_transform
            )
        if stage == "predict":
            self.val_cfd_datamodule = create_cfd_datamodule(
                self.metadata_files, self.batch_sizes, self.loader_types, self.start_idx,
                shuffle=False, split=self.split, flag='val',
                collater_transform=self.graph_transform
            )

    def prepare_data(self):
        None

    def train_dataloader(self):
        return self.train_cfd_datamodule.get_combined_loader()

    def val_dataloader(self):
        return self.val_cfd_datamodule.get_combined_loader(mode='sequential')

    def test_dataloader(self):
        return self.val_cfd_datamodule.get_combined_loader(mode='sequential')

    def predict_dataloader(self):
        return self.val_cfd_datamodule.get_combined_loader(mode='sequential')
```

```python
metadata_files = [
        os.path.join('/scratch/coop/theret/cfd-dataset/tutorial/sample_dataset/metadata.yaml')
   ]

# Diffusion process
diffusion_process = DiffusionProcess(
    num_steps     = 1000,
    schedule_type = 'linear',
)

# Model
arch = {
    'in_node_features':   3,
    'cond_node_features': 0,
    'cond_edge_features': 3,
    'depths':             [3,3,3,3],
    'fnns_width':         128,
    'aggr':               'sum',
    'dropout':            0.1,
    'dim':                  3,
    "scalar_rel_pos":   False
}
net = DiffusionGraphNet(
    diffusion_process  = diffusion_process,
    learnable_variance = True,
    arch               = arch,
)

# Loss and sampler
criterion = HybridLoss()              # (model, graph) -> [B]
step_sampler_factory = ImportanceStepSampler

# LightningModule wrapper
lit = LitDiffusionCFD(
    net=net,
    diffusion_process=diffusion_process,
    criterion=criterion,
    step_sampler_factory=step_sampler_factory,
    lr=1e-4,
    scheduler_cfg={"factor":0.1, "patience":50},
    pack_mode="y_window_cond_static",
    pack_win_len=1,            # <— use these names
    pack_stride=1,
    pack_select="random",
    y_idx=[0,1,2],
    cond_idx=None,
)


# DataModule from your CFDDataset
dm = cfd_datamodule(metadata_files, train_val_split=0.8)
```

```python

# Trainer
prog_bar = RichProgressBar(theme=RichProgressBarTheme(description="green_yellow", 
                                                      progress_bar="green1", 
                                                      progress_bar_finished="green1", 
                                                      progress_bar_pulse="#6206E0", 
                                                      batch_progress="green_yellow", 
                                                      time="grey82", 
                                                      processing_speed="grey82", 
                                                      metrics="grey82", 
                                                      metrics_text_delimiter="\n", 
                                                      metrics_format=".3e"))

ckpt = ModelCheckpoint(dirpath="checkpoints", 
                       filename="diffusion-{epoch}", 
                       monitor="val/loss", 
                       mode="min", 
                       save_top_k=3,
                       save_last=True)

trainer = L.Trainer(max_epochs=2, 
                    accelerator="auto", 
                    precision="16-mixed", 
                    callbacks=[ckpt, prog_bar], 
                    log_every_n_steps=10, 
                    limit_val_batches=20, 
                    limit_train_batches=80,
                    accumulate_grad_batches=64)

# Train
# after you construct `trainer`, `lit`, and `dm`:
ckpt_path = "/scratch/coop/theret/nn4avbp/checkpoints/last.ckpt"  # or a specific epoch file like "checkpoints/diffusion-epoch=1.ckpt"
if ckpt_path is None:
    trainer.fit(lit, dm) 
else:
    trainer.fit(lit, dm, ckpt_path=ckpt_path)
```

```python
# ---- quick_plot.py (paste anywhere in your project) ----
import numpy as np
import torch
import matplotlib.pyplot as plt

def quick_plot_pred(out, dm, graph_id=0, comp="mag", quiver=True, save=None):
    """
    Super-simple plotter for LitDiffusionCFD.predict() outputs.
    - out: dict returned by predict_step (must include 'pred_norm','pos','batch')
    - dm:  your cfd_datamodule (for Z-score mean/std)
    - graph_id: which graph in the batch to visualize
    - comp: 'u'/'v'/'w'/0/1/2 for a component, or 'mag' for magnitude
    - quiver: overlay a decimated quiver for (u,v)
    - save: optional path to save PNG
    """
    # Tensors -> numpy and denorm
    pred_norm = out["pred_norm"]
    pos       = out["pos"]
    batch_idx = out["batch"]
    if isinstance(pred_norm, torch.Tensor): pred_norm = pred_norm.cpu().numpy()
    if isinstance(pos, torch.Tensor):       pos       = pos.cpu().numpy()
    if isinstance(batch_idx, torch.Tensor): batch_idx = batch_idx.cpu().numpy()

    mean = dm._zscore_mean.cpu().numpy()
    std  = dm._zscore_std.cpu().numpy()
    pred = pred_norm * std + mean  # [N, C]

    # Select a single graph
    m = (batch_idx == graph_id)
    xy = pos[m, :2]
    x, y = xy[:, 0], xy[:, 1]

    # Pick component to color
    if comp in ("u", 0):   z = pred[m, 0]
    elif comp in ("v", 1): z = pred[m, 1]
    elif comp in ("w", 2): z = pred[m, 2]
    else:                  z = np.linalg.norm(pred[m, :3], axis=1)  # 'mag'

    # Base scatter heatmap
    plt.figure(figsize=(6,5), dpi=120)
    sc = plt.scatter(x, y, c=z, s=8)
    plt.colorbar(sc, label=str(comp))
    plt.gca().set_aspect('equal', 'box')
    plt.xlabel("x"); plt.ylabel("y"); plt.title(f"Prediction: {comp}")

    # Optional quiver (uses u,v)
    if quiver:
        U, V = pred[m, 0], pred[m, 1]
        # light decimation so it stays readable
        step = max(len(x)//150, 1)
        plt.quiver(x[::step], y[::step], U[::step], V[::step],
                   angles="xy", scale_units="xy", scale=None, width=0.002, alpha=0.8)

    plt.tight_layout()
    if save: plt.savefig(save, bbox_inches="tight")
    plt.show()
```

```python
# ---- quick_plot_3d.py ----
import numpy as np
import torch
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # registers 3D projection

def _axes_equal_3d(ax):
    # make 3D axes have equal scale
    xlim = np.array(ax.get_xlim3d())
    ylim = np.array(ax.get_ylim3d())
    zlim = np.array(ax.get_zlim3d())
    xmid, xrad = xlim.mean(), (xlim[1]-xlim[0]) / 2
    ymid, yrad = ylim.mean(), (ylim[1]-ylim[0]) / 2
    zmid, zrad = zlim.mean(), (zlim[1]-zlim[0]) / 2
    R = max(xrad, yrad, zrad)
    ax.set_xlim3d([xmid - R, xmid + R])
    ax.set_ylim3d([ymid - R, ymid + R])
    ax.set_zlim3d([zmid - R, zmid + R])

def quick_plot_pred_3d(out, dm, graph_id=0, comp="mag", quiver=True, step=None, save=None):
    """
    3D visualization for LitDiffusionCFD.predict() outputs.
    - out: dict with 'pred_norm','pos','batch' from predict_step
    - dm:  your datamodule (for Z-score mean/std)
    - comp: 'u'/'v'/'w'/0/1/2 or 'mag' (color)
    - quiver: draw 3D arrows (u,v,w). Set False to disable.
    - step: decimation factor for quiver (auto if None)
    """
    pred_norm = out["pred_norm"]
    pos       = out["pos"]
    batch_idx = out["batch"]
    if torch.is_tensor(pred_norm): pred_norm = pred_norm.detach().cpu().numpy()
    if torch.is_tensor(pos):       pos       = pos.detach().cpu().numpy()
    if torch.is_tensor(batch_idx): batch_idx = batch_idx.detach().cpu().numpy()

    # denormalize to physical units
    #mean = dm._zscore_mean.detach().cpu().numpy()
    #std  = dm._zscore_std.detach().cpu().numpy()
    #pred = pred_norm * std + mean  # [N, C]
    pred = pred_norm
    # pick one graph
    m = (batch_idx == graph_id)
    P = pos[m]
    x = P[:, 0]
    y = P[:, 1]
    z = P[:, 2] if P.shape[1] >= 3 else np.zeros_like(x)  # pad Z if 2D mesh

    # vector components
    U, V = pred[m, 0], pred[m, 1]
    W = pred[m, 2] if pred.shape[1] >= 3 else np.zeros_like(U)

    # color by requested component
    if comp in ("u", 0):   color_vals = U
    elif comp in ("v", 1): color_vals = V
    elif comp in ("w", 2): color_vals = W
    else:                  color_vals = np.sqrt(U**2 + V**2 + W**2)  # 'mag'

    fig = plt.figure(figsize=(7,6), dpi=120)
    ax = fig.add_subplot(111, projection='3d')

    sc = ax.scatter(x, y, z, c=color_vals, s=6)
    cb = fig.colorbar(sc, ax=ax, shrink=0.7)
    cb.set_label(str(comp))

    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_title(f"Prediction (3D): {comp}")

    # optional 3D quiver
    if quiver:
        n = len(x)
        if step is None:
            step = max(n // 400, 1)  # light decimation for readability
        # pick a reasonable arrow length based on data extent
        span = max(x.max()-x.min(), y.max()-y.min(), z.max()-z.min())
        length = 0.04 * span if np.isfinite(span) and span > 0 else 1.0
        ax.quiver(x[::step], y[::step], z[::step],
                  U[::step], V[::step], W[::step],
                  length=length, normalize=False, linewidths=0.5, alpha=0.8)

    # equal aspect for 3D
    _axes_equal_3d(ax)
    plt.tight_layout()
    if save: plt.savefig(save, bbox_inches="tight")
    plt.show()
```

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa

def _axes_equal_3d(ax):
    xlim = np.array(ax.get_xlim3d()); ylim = np.array(ax.get_ylim3d()); zlim = np.array(ax.get_zlim3d())
    xmid, xrad = xlim.mean(), (xlim[1]-xlim[0]) / 2
    ymid, yrad = ylim.mean(), (ylim[1]-ylim[0]) / 2
    zmid, zrad = zlim.mean(), (zlim[1]-zlim[0]) / 2
    R = max(xrad, yrad, zrad)
    ax.set_xlim3d([xmid - R, xmid + R])
    ax.set_ylim3d([ymid - R, ymid + R])
    ax.set_zlim3d([zmid - R, zmid + R])

def quick_plot_pred_3d(
    out, dm, graph_id=0, comp="mag", quiver=True, step=None, save=None,
    # NEW controls:
    vec_clip=None,                 # float or (umin, umax) tuple to clip U (applies to all if float)
    u_range=None, v_range=None, w_range=None,  # per-component (min,max) clip
    color_range=None,              # (vmin, vmax) for colormap of the chosen component
    quiver_normalize=False,        # True = all arrows same length (direction only)
    quiver_length_factor=0.04,     # base arrow length as fraction of domain span
    vec_scale=None,                # multiply U,V,W by this scalar (e.g., to bring to common scale across plots)
):
    pred_norm = out["pred_norm"]; pos = out["pos"]; batch_idx = out["batch"]
    if torch.is_tensor(pred_norm): pred_norm = pred_norm.detach().cpu().numpy()
    if torch.is_tensor(pos):       pos       = pos.detach().cpu().numpy()
    if torch.is_tensor(batch_idx): batch_idx = batch_idx.detach().cpu().numpy()

    # If you want physical units, uncomment:
    mean = dm._zscore_mean.detach().cpu().numpy()
    std  = dm._zscore_std.detach().cpu().numpy()
    pred = pred_norm * std + mean
    #pred = pred_norm
    pred = pred_norm
    m = (batch_idx == graph_id)
    P = pos[m]
    x, y = P[:, 0], P[:, 1]
    z = P[:, 2] if P.shape[1] >= 3 else np.zeros_like(x)

    U, V = pred[m, 0].copy(), pred[m, 1].copy()
    W = pred[m, 2].copy() if pred.shape[1] >= 3 else np.zeros_like(U)

    # ---- vector scaling/capping ----
    # Global symmetric clip for all components
    if isinstance(vec_clip, (int, float)) and vec_clip is not None:
        U = np.clip(U, -vec_clip, vec_clip)
        V = np.clip(V, -vec_clip, vec_clip)
        W = np.clip(W, -vec_clip, vec_clip)
    # Per-component ranges (override if provided)
    if u_range is not None: U = np.clip(U, u_range[0], u_range[1])
    if v_range is not None: V = np.clip(V, v_range[0], v_range[1])
    if w_range is not None: W = np.clip(W, w_range[0], w_range[1])
    # Scale vectors (e.g., to harmonize across runs)
    if vec_scale is not None:
        U *= vec_scale; V *= vec_scale; W *= vec_scale

    # ---- color selection & range ----
    if comp in ("u", 0):   color_vals = U
    elif comp in ("v", 1): color_vals = V
    elif comp in ("w", 2): color_vals = W
    else:                  color_vals = np.sqrt(U**2 + V**2 + W**2)  # 'mag'

    vmin = vmax = None
    if color_range is not None:
        vmin, vmax = color_range

    fig = plt.figure(figsize=(7,6), dpi=120)
    ax = fig.add_subplot(111, projection='3d')

    sc = ax.scatter(x, y, z, c=color_vals, s=6, vmin=vmin, vmax=vmax)
    cb = fig.colorbar(sc, ax=ax, shrink=0.7); cb.set_label(str(comp))

    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_title(f"Prediction (3D): {comp}")

    if quiver:
        n = len(x)
        if step is None:
            step = max(n // 400, 1)
        span = max(x.max()-x.min(), y.max()-y.min(), z.max()-z.min())
        base_len = quiver_length_factor * span if np.isfinite(span) and span > 0 else 1.0

        # If normalize=True: all arrows have the same length (base_len)
        ax.quiver(
            x[::step], y[::step], z[::step],
            U[::step], V[::step], W[::step],
            length=base_len, normalize=bool(quiver_normalize),
            linewidths=0.5, alpha=0.8
        )

    _axes_equal_3d(ax)
    plt.tight_layout()
    if save: plt.savefig(save, bbox_inches="tight")
    plt.show()
```

```python
# --- One-cell: plot one *training* sample (GT) WITHOUT ARROWS ---

import numpy as np
import torch
import matplotlib.pyplot as plt
import matplotlib.tri as mtri

# If dm doesn't exist yet, build it
try:
    dm
except NameError:
    dm = cfd_datamodule(metadata_files, train_val_split=0.8)

# Prepare stats/transforms
dm.setup("fit")

# --- get one batch from the *training* loader (already transformed & coarsened) ---
dl = dm.train_dataloader()
item = next(iter(dl))  # CombinedLoader item

def _get_sequence_from_combined(item):
    # item may be (batches, batch_idx, dataloader_idx) OR just batches
    if isinstance(item, tuple) and len(item) == 3:
        batches, _, _ = item
    else:
        batches = item
    if isinstance(batches, dict):
        sequence = next(iter(batches.values()))
    else:
        sequence = batches
    return list(sequence) if isinstance(sequence, tuple) else sequence

sequence = _get_sequence_from_combined(item)
g = sequence[-1]  # last timestep (PyG Batch), already through dm.graph_transform

# --- helpers ---
def _to_np(x):
    return x.detach().cpu().numpy() if torch.is_tensor(x) else np.asarray(x)

def _remap_triangles_to_mask(tri, mask_bool):
    """Map global triangles (M,3) to compact indices of nodes where mask_bool==True."""
    if tri is None:
        return None
    tri = _to_np(tri)
    if tri.ndim != 2 or tri.shape[1] != 3:
        return None
    mask_np = _to_np(mask_bool).astype(bool)
    if mask_np.sum() < 3:
        return None
    old2new = -np.ones(mask_np.shape[0], dtype=np.int64)
    old2new[mask_np] = np.arange(mask_np.sum(), dtype=np.int64)
    keep = mask_np[tri].all(axis=1)
    if not keep.any():
        return None
    return old2new[tri[keep]]

def _axes_equal_3d(ax):
    xlim = np.array(ax.get_xlim3d()); ylim = np.array(ax.get_ylim3d()); zlim = np.array(ax.get_zlim3d())
    xmid, ymid, zmid = xlim.mean(), ylim.mean(), zlim.mean()
    R = max((xlim[1]-xlim[0])/2, (ylim[1]-ylim[0])/2, (zlim[1]-zlim[0])/2)
    ax.set_xlim3d([xmid-R, xmid+R]); ax.set_ylim3d([ymid-R, ymid+R]); ax.set_zlim3d([zmid-R, zmid+R])

# --- denormalize GT ---
mean = dm._zscore_mean.cpu()
std  = dm._zscore_std.cpu()

pos   = g.pos.cpu()
batch = g.batch.cpu()
gt    = g.target.cpu() * std + mean  # [N, C]

graph_id = 0
m = (batch == graph_id)

P = pos[m]
x, y = P[:, 0].numpy(), P[:, 1].numpy()
z = P[:, 2].numpy() if P.shape[1] >= 3 else np.zeros_like(x)

C = gt.shape[1]
U = gt[m, 0].numpy() if C >= 1 else np.zeros_like(x)
V = gt[m, 1].numpy() if C >= 2 else np.zeros_like(x)
W = gt[m, 2].numpy() if C >= 3 else np.zeros_like(x)
mag = np.sqrt(U**2 + V**2 + W**2)

# triangles (if present) remapped to the masked subgraph
tri_all = None
if hasattr(g, "cells") and isinstance(g.cells, list) and g.cells and g.cells[0] is not None:
    tri_all = g.cells[0]  # could be np.array or tensor
triangles = _remap_triangles_to_mask(tri_all, m)

# --- plot (auto 2D/3D), NO ARROWS ---
if P.shape[1] == 2:
    plt.figure(figsize=(6,5), dpi=120)
    if triangles is not None:
        tri = mtri.Triangulation(x, y, triangles)
        tcf = plt.tricontourf(tri, mag, levels=64)
        plt.triplot(tri, lw=0.2, alpha=0.15)
        plt.colorbar(tcf, label="|u|")
    else:
        sc = plt.scatter(x, y, c=mag, s=8, edgecolors="none")
        plt.colorbar(sc, label="|u|")
    plt.gca().set_aspect("equal", "box")
    plt.xlabel("x"); plt.ylabel("y"); plt.title("GT |u| (one training sample)")
    plt.tight_layout(); plt.show()
else:
    from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
    fig = plt.figure(figsize=(7,6), dpi=120)
    ax = fig.add_subplot(111, projection='3d')
    sc = ax.scatter(x, y, z, c=mag, s=6, edgecolors="none")
    fig.colorbar(sc, ax=ax, shrink=0.7).set_label("|u|")
    for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
        try: axis._axinfo["grid"]["linewidth"] = 0.0
        except: pass
        if hasattr(axis, "pane"): axis.pane.set_visible(False)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_title("GT |u| (one training sample, 3D)")
    _axes_equal_3d(ax)
    plt.tight_layout(); plt.show()
```

```python
# --- choose which components to plot ---
COMPS = ["u", "v", "w", "mag"]   # if your data is 2D (no w), this will auto-skip "w"

# Optional: lock color ranges for fair comparison across plots
LOCK_RANGE = True

# Build a dict of fields
fields = {
    "u": U,
    "v": V,
    "w": W,
    "mag": mag,
}

# If there is no third component in your data, drop "w"
if gt.shape[1] < 3:
    COMPS = [c for c in COMPS if c != "w"]

# Helper to pick color limits
def color_limits(name, arr):
    if not LOCK_RANGE:
        return None, None
    arr = np.asarray(arr)
    if name == "mag":
        return 0.0, float(np.nanmax(arr))  # magnitude is non-negative
    vmax = float(np.nanmax(np.abs(arr)))   # symmetric around 0 for u/v/w
    return -vmax, vmax

for comp in COMPS:
    color_vals = fields[comp]

    if P.shape[1] == 2:
        plt.figure(figsize=(6,5), dpi=120)
        vmin, vmax = color_limits(comp, color_vals)
        if triangles is not None:
            tri = mtri.Triangulation(x, y, triangles)
            tcf = plt.tricontourf(tri, color_vals, levels=64, vmin=vmin, vmax=vmax)
            plt.triplot(tri, lw=0.2, alpha=0.15)
            plt.colorbar(tcf, label=comp)
        else:
            sc = plt.scatter(x, y, c=color_vals, s=8, edgecolors="none", vmin=vmin, vmax=vmax)
            plt.colorbar(sc, label=comp)
        plt.gca().set_aspect("equal", "box")
        plt.xlabel("x"); plt.ylabel("y"); plt.title(f"GT {comp} (one training sample)")
        plt.tight_layout(); plt.show()
    else:
        from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
        fig = plt.figure(figsize=(7,6), dpi=120)
        ax = fig.add_subplot(111, projection='3d')
        vmin, vmax = color_limits(comp, color_vals)
        sc = ax.scatter(x, y, z, c=color_vals, s=6, edgecolors="none", vmin=vmin, vmax=vmax)
        fig.colorbar(sc, ax=ax, shrink=0.7).set_label(comp)
        for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
            try: axis._axinfo["grid"]["linewidth"] = 0.0
            except: pass
            if hasattr(axis, "pane"): axis.pane.set_visible(False)
        ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
        ax.set_title(f"GT {comp} (one training sample, 3D)")
        _axes_equal_3d(ax)
        plt.tight_layout(); plt.show()
```
